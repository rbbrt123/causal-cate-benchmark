```python
import numpy as np
```

```python
# number of samples
n_samples = 1000

# treatment probability
p_treatment = 0.05
```

# Define rng

```python
rng = np.random.default_rng(42)
```

# Generate covariates

```python
X = rng.normal(0, 1, (n_samples,2))
```

```python
X
```

# Generate treatments

```python
T = rng.binomial(1, p_treatment, n_samples)
```

```python
T
```

# Generate mu0, mu1 and tau

```python
# these are deterministic functions

def generate(x1, x2):
    mu0 = 2 + x1 + 0.5*x2
    tau = 1 + 0.8*x1
    mu1 = mu0 + tau

    return mu0, mu1, tau
```

```python
mu0, mu1, tau = generate(X[:,0], X[:,1])
```

# Generate the observed outcomes

```python
# draw noise
epsilon = rng.normal(0, 1, n_samples)
```

```python
def generate_observed_outcomes(mu0, tau, T, epsilon):
    return mu0 + (T*tau) + epsilon
```

```python
Y = generate_observed_outcomes(mu0, tau, T, epsilon)
```
