# Define rng

# Generate mu0, mu1 and tau

```python
def generate(x1, x2):
    mu0 = 2 + x1 + 0.5*x2
    tau = 1 + 0.8*x1
    mu1 = mu0 + tau

    return mu0, mu1, tau
```

```python
mu0, mu1, tau = generate(X[:,0], X[:,1])
```
