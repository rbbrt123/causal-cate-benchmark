```python
from econml.metalearners import TLearner
from cate_benchmark.data_generator import generate_dataset
from cate_benchmark.evaluation import root_pehe
from sklearn.ensemble import RandomForestRegressor
```

```python
X_train, T_train, Y_train, tau_train = generate_dataset(1000, 0.5, 42)
X_test, T_test, Y_test, tau_test = generate_dataset(500, 0.5, 43)
```

```python
print(X_train.shape)
print(T_train.shape)
print(Y_train.shape)
print(X_test.shape)
```

# T-learner

```python
# Create baselearner
base_learner = RandomForestRegressor(random_state=42)

# Create T-learner
T_learner = TLearner(models=base_learner)

# Fit T-learner
T_learner.fit(Y_train, T_train, X=X_train)

# Estimate CATE
estimated_cate_econml = T_learner.effect(X_test)
```

```python
pehe_t_econml = root_pehe(estimated_cate_econml, tau_test)
```
